```python
import json
import pyperclip
import pandas as pd
import numpy as np
from pprint import pprint
```

```python
df = pd.read_clipboard()
peaks = {}
for i,b in enumerate(df["Checked"]):
    if b == True:
        hkl = df["hkl"][i]
        tth = float(df["Observed X"][i])
        peaks[hkl] = tth
text = json.dumps(peaks, indent = 4)
print(text)
pyperclip.copy(text)
```

```python
s = pyperclip.paste()
df = pd.DataFrame(np.array(s.split(sep = "\t"))[:-1].reshape(-1,2))
df.columns = ["Value", "error"]
df.index = ["a", "b", "c", "alpha", "beta", "gamma", "volume"]
for k in df.keys():
    df[k] = pd.to_numeric(df[k], errors = "coerce")
df = df.dropna()
print(df)
text = json.dumps(df.T.to_dict(), indent = 4)
pyperclip.copy(text)
```
