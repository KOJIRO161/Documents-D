連続データを指定します。  
基本的にフレーム番号が末尾に書いてあるはずなので、ディレクトリとフレーム番号以外を指定して、連続データを決めてやります。

```python
# ディレクトリ名を指定
dir = r"D:\DATA\SPring-8-2025-Nov\Okazaki\UODE32\XRD1"

# ヘッダーとフッターを指定
header = "UODE32_1_"
footer = ".his"
```

```python
# ディレクトリの中からヘッダーとフッターを含むファイル名を取得
flist = list()
for __ in os.listdir(dir):
    if not header in __:
        continue
    if not footer in __:
        continue
    flist.append(__)
print("Directory: " + os.path.abspath(dir))
print("Header:    " + header)
print("Footer:    " + footer)

# ソート
flist.sort(key = (lambda x: int(x.replace(header, "").replace(footer, ""))))
filenames = list(map(lambda x: os.path.join(dir, x), flist))

# 表示
for f in filenames:
    print(f)

# 格納
i2a.filelist = filenames
```
