## 初期設定

```python
import sys, os
from IPython.display import display, HTML, clear_output, update_display, Image

# 自作モジュール
sys.path.append(r"C:\Users\okaza\pythonenv")
from modules.Mytools.handle_ipynb import save_pickle, load_pickle, export_html
sys.path.append(os.getcwd())
from main import Series2img # type: ignore

# 初期パラメーター
cachedir = os.path.join(os.getcwd(), ".cache")
os.makedirs(cachedir, exist_ok=True)
```

## ファイル読み込み

```python
# runnameを指定
run_number = 32

# runnameを設定
runname = "UODE29_{:0>4}".format(run_number)

# ディレクトリ名を指定
dir = r"D:\SPring-8-2025-Oct\Okazaki\UODE29\XRD2"

# ヘッダーとフッターを指定
header = "UODE29_FPD_{}_".format(run_number)
footer = ".his"

si.his2hdf(dir = dir,
           header = header,
           footer = footer,
           initfilename=runname+"_raw")
```

## 切り開き

```python
# poniファイル
poni = r"D:\SPring-8-2025-Oct\ceo2_20251003\CeO2_FPD_MgS400_20251003_2.poni"

# 切り開きの細かさ
rad = 4096

si.caking(poni = poni,
          rad = rad,
          initfilename = runname + "_caking")
```

## 時系列画像出力

```python
# 縦軸の範囲
theta_lim = (8,20)
theta_lim = (12,14)
# # theta_lim = (12.9,13.1)

# pseudo Voigt関数でフィッティングしたピークを示す (optional)
# fit = "GREEN"

# フレームレート
fps = 0.563954695532576 # Hz
fps = 1 # Hz

imgfile = si.imshow_1dseries(theta_lim=theta_lim,
                             fps = fps,
                             fit = True if ("fit" in locals()) else False)
display(Image(filename = imgfile, width = 600))
```
